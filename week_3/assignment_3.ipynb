## 1.Find all orders where sales are greater than the average sales. (Subquery)  

```python
pd.read_sql_query( "SELECT AVG(sales) FROM order_details",conn)
```

```python

pd.read_sql_query( """SELECT *,
                              (SELECT AVG(sales) FROM order_details) as avg_sales
                      FROM order_details 
                      WHERE sales > (
                                      SELECT AVG(sales) 
                                      FROM order_details
                                    )""",conn)
```

The query uses a subquery in the WHERE clause to calculate the average sales and returns all orders whose sales are greater than that average.

## 2.Find the highest sales order for each customer. (Subquery)  

```python
pd.read_sql_query("""
                SELECT 
                c.cust_name,     t1.cust_id,        t1.order_id as highest_sales_order ,     t1.order_sales
                FROM
                (
                    SELECT
                    o.cust_id,          o.order_id,         SUM(od.sales) AS order_sales
                    FROM orders o
                    JOIN order_details od
                    ON o.order_id = od.order_id
                    GROUP BY o.cust_id, o.order_id
                )
                t1
                JOIN customers c
                ON c.cust_id = t1.cust_id
                WHERE order_sales =
                (
                    SELECT MAX(t2.order_sales)
                    FROM
                    (
                        SELECT 
                        o.cust_id,       o.order_id,          SUM(od.sales) AS order_sales
                        FROM orders o
                        JOIN order_details od
                        ON o.order_id = od.order_id
                        GROUP BY o.cust_id, o.order_id
                    ) 
                    t2
                    WHERE t2.cust_id = t1.cust_id
                )
                """,conn)
```

Execution Order:

- FROM : Creates table t1 containing cust_id, order_id and total sales for each order.
- WHERE : For each row of t1, correlated subquery t2 finds the maximum order_sales for that customer using t2.cust_id = t1.cust_id.
- WHERE : Retains only those rows whose order_sales equals the maximum order_sales for that customer.
- SELECT : Displays cust_name, cust_id, highest_sales_order_id and the corresponding order_sales.

## 3.Calculate total sales for each customer. (CTE)  

```python
pd.read_sql_query("""
                    WITH cust_sales AS
                    (
                        SELECT 
                        o.cust_id,       SUM(od.sales) AS total_sales
                        FROM orders o
                        JOIN order_details od
                          ON o.order_id = od.order_id
                        GROUP BY o.cust_id
                    )
                    
                    SELECT 
                    c.cust_name,      cs.cust_id,            cs.total_sales
                    FROM cust_sales cs
                    JOIN customers c
                      ON cs.cust_id = c.cust_id
                    """,conn)
```


CTE query=>
- FROM : Joins orders and order_details using order_id to associate each sale with a customer.
- GROUP BY : Groups records by cust_id and calculates total_sales for each customer using SUM(sales).
- WITH(CTE) : Creates a temporary result set named cust_sales containing cust_id anad total_sales.

Main query =>
- FROM : Joins the CTE cust_sales with customers using cust_id.
- SELECT : Displays cust_name, cust_id and total_sales.

## 4.Find customers whose total sales are above average. (CTE + Subquery)  

```python
pd.read_sql_query("""
                    WITH cust_sales AS
                    (
                        SELECT
                        o.cust_id,           SUM(od.sales) AS total_sales
                        FROM orders o
                        JOIN order_details od
                          ON o.order_id = od.order_id
                        GROUP BY o.cust_id
                    )
                    
                    SELECT
                    c.cust_name,    cs.cust_id,           cs.total_sales
                    FROM cust_sales cs
                    JOIN customers c
                                  ON cs.cust_id = c.cust_id
                    WHERE cs.total_sales >
                                            (
                                                SELECT AVG(total_sales)
                                                FROM cust_sales
                                            )
                    """,conn)
    
```

CTE Quer=>
- FROM : Joins orders and order_details using order_id to associate each sale with a customer.
- GROUP BY : Groups records by cust_id and calculates total_sales for each customer using SUM(sales).
- WITH(CTE) :Temporary set containing cust_id and total_sales.

Main Query=>
- FROM : Joins the CTE with customers using cust_id.
- WHERE : Uses a subquery to calculate the aveg total_sales of all customers from CTE and retains only customers whose tot_sales is greater than average.
- SELECT : Displays cust_name, cust_id and total_sales of customers whose total sales are above average.

## 5. Rank all customers based on total sales. (Window Function)

```python
pd.read_sql_query(
                """
                WITH cust_sales AS
                (
                    SELECT
                                o.cust_id,         SUM(od.sales) AS tot_sales
                    FROM orders o
                    JOIN order_details od
                                    ON o.order_id = od.order_id
                    GROUP BY o.cust_id
                )
                
                SELECT
                    c.cust_name,             cs.cust_id,               cs.tot_sales,            
                    RANK() OVER (ORDER BY cs.tot_sales DESC) AS rank
                FROM cust_sales cs
                JOIN customers c  
                                ON cs.cust_id = c.cust_id;
                """,conn)
```

CTE Query=>
- FROM : Joins orders and order_details using order_id to associate each sale with a customer.
- GROUP BY : Groups records by cust_id and calculates tot_sales for each customer using SUM(sales).
- WITH (CTE) : Creates a temporary set named cust_sales containing cust_id and tot_sales.

Main Query=>
- FROM : Joins the CTE cust_sales with customers using cust_id.
- SELECT : Retrieves cust_name, cust_id and tot_sales.
- WINDOW FUNCTION : Used RANK() to assign ranks to customers based on total sales.

## 6.Assign row numbers to each order within a customer. (Window Function + PARTITION BY)  

```python
pd.read_sql_query(
                """
                SELECT
                        o.cust_id,             o.order_id,             o.order_date,
                        ROW_NUMBER() OVER
                        (
                            PARTITION BY o.cust_id
                            ORDER BY o.order_date
                        ) AS row_num
                FROM orders o;
                """,conn)
```

- PARTITION BY o.cust_id creates separate groups for each customer.
- ORDER BY o.order_date arranges the customer's orders by date before assigning numbers.
- ROW_NUMBER() gives a unique sequential number 1, 2, 3... to each order within that customer group.

## 7.Display top 3 customers based on total sales. (Window Function)  

```python
pd.read_sql_query(
                """
                WITH cust_sales AS
                (
                    SELECT
                            o.cust_id,         SUM(od.sales) AS tot_sales
                    FROM orders o
                    JOIN order_details od 
                                        ON o.order_id = od.order_id
                    GROUP BY o.cust_id
                )

                SELECT
                    c.cust_name,             cs.cust_id,             cs.tot_sales
                FROM
                (
                    SELECT
                            cust_id,             tot_sales,
                        ROW_NUMBER() OVER (ORDER BY tot_sales DESC) AS rn
                    FROM cust_sales
                ) cs
                JOIN customers c 
                                ON cs.cust_id = c.cust_id
                WHERE cs.rn<=3;
                """,conn)
```

- First, total sales are calculated for each customer using SUM(sales).
- ROW_NUMBER() ranks customers from highest to lowest sales.
- WHERE rn<=3 keeps only the top 3 customers with the highest sales.

Answer the following using SQL: 

## 1. Who are the top 5 customers?

## 2.Who are the bottom 5 customers?

## 3.Which customers made only one order?

```python
pd.read_sql_query(
                """
                SELECT
                    c.cust_name,             o.cust_id ,
                    COUNT(o.order_id) AS tot_orders
                FROM orders o
                JOIN customers c 
                                ON o.cust_id= c.cust_id
                GROUP BY 
                                c.cust_name         , o.cust_id
                HAVING COUNT(o.order_id) =1;
                """,conn)
```

- COUNT(order_id) counts how many orders each customer placed.
- GROUP BY creates one row per customer.

# 4.Which customers have above-average sales?

```python
pd.read_sql_query(
                """
                WITH cust_sales AS
                (
                    SELECT
                        o.cust_id,         SUM(od.sales) AS tot_sales
                    FROM orders o
                    JOIN order_details od ON o.order_id = od.order_id
                    GROUP BY o.cust_id
                )

                SELECT
                    c.cust_name,             cs.tot_sales,
                    (SELECT AVG(tot_sales) FROM cust_sales) AS avg_sales
                FROM cust_sales cs
                JOIN customers c ON cs.cust_id = c.cust_id
                WHERE cs.tot_sales >
                (
                    SELECT AVG(tot_sales)
                    FROM cust_sales
                );
                """,conn)
```

- cust_sales calculates the total sales for each customer.
- The subquery AVG(tot_sales) finds the avg sales of all customers.

## 5.What is the highest order value per customer?

```python
pd.read_sql_query(
                """
                WITH ord_sales AS
                (
                    SELECT
                        o.cust_id,         o.order_id,
                        SUM(od.sales) AS ord_value
                    FROM orders o
                    JOIN order_details od 
                                         ON o.order_id = od.order_id
                    GROUP BY o.cust_id, o.order_id
                )

                SELECT
                    c.cust_name ,             os.cust_id,
                    MAX(os.ord_value) AS highest_order_value
                FROM ord_sales os
                JOIN customers c 
                                ON os.cust_id= c.cust_id
                GROUP BY 
                            c.cust_name, os.cust_id;
                """,conn)
```

- First, the total value of each order is calculated.
- MAX() finds the highest order value for each customer.
